### 🔹 Ejercicio 10: Conversión de Tipos (Type Casting)
**Nivel:** Intermedio

**Objetivo:** Corregir el `TypeError` del Ejercicio 5 y forzar la conversión de tipos (casting).

**Consigna:**
1. Crea una variable `etiqueta = "Edad: "`.
2. Crea una variable `edad_num = 30`.
3. Intenta crear `mensaje = etiqueta + edad_num` (esto dará un error).
4. Corrige el error convirtiendo `edad_num` a un string usando `str()` antes de concatenar.
5. Almacena el resultado correcto en `mensaje_correcto`.

```python
# Tu código aquí

etiqueta = "Edad: "
edad_num = 30

# Corrección usando str() para convertir el int a str
mensaje_correcto = etiqueta + str(edad_num)

print(mensaje_correcto)
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---

etiqueta = "Edad: "
edad_num = 30

# Corrección usando str() para convertir el int a str
mensaje_correcto = etiqueta + str(edad_num)

print(mensaje_correcto)
```

* `float(valor)` convierte `valor` a flotante (ej. `30` -> `30.0` o `"30.5"` -> `30.5`).

**Variante más optimizada (f-strings):**
La forma moderna de hacer esto es usar f-strings, que hacen la conversión automáticamente:
`mensaje_fstring = f"Edad: {edad_num}"`
Esto es lo que hemos estado usando en los `print()`.

### 🔹 Ejercicio 13: Intercambio de Variables (Swapping)
**Nivel:** Intermedio

**Objetivo:** Intercambiar los valores de dos variables (un uso común de la asignación múltiple).

**Consigna:**
1. Crea `x = 10` e `y = 20`.
2. Sin usar una tercera variable temporal (como `temp`), intercambia los valores de `x` e `y`.
3. Imprime `x` (debería ser 20) e `y` (debería ser 10).

```python
# Tu código aquí
x = 10
y = 20

x, y = y, x

print(f"x ahora es: {x}")
print(f"y ahora es: {y}")
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---
x = 10
y = 20

print(f"Valores iniciales: x={x}, y={y}")

# Intercambio (Swapping) usando asignación múltiple
x, y = y, x

print(f"Valores intercambiados: x={x}, y={y}")
```

**Explicación:**
* `x, y = y, x` es la forma "Pythónica" de intercambiar variables.
* Python primero evalúa la expresión de la derecha: `(y, x)` que es `(20, 10)`.
* Luego, asigna (desempaqueta) esa tupla a las variables de la izquierda `x, y`.
* En otros lenguajes, esto requeriría una variable temporal: `temp = x; x = y; y = temp`.

### 🔹 Ejercicio 16: División Entera y Módulo
**Nivel:** Intermedio

**Objetivo:** Entender los operadores `//` (división entera) y `%` (módulo/resto).

**Consigna:**
Quieres repartir 17 caramelos entre 5 niños.
1. Crea `caramelos = 17` y `niños = 5`.
2. Calcula cuántos caramelos enteros recibe cada niño (división entera `//`) y guárdalo en `caramelos_por_niño`.
3. Calcula cuántos caramelos sobran (módulo `%`) y guárdalo en `caramelos_sobrantes`.

```python
# Tu código aquí
caramelos = 17
niños = 5

caramelos_por_niño = caramelos // niños
caramelos_sobrantes = caramelos % niños

print(f"Caramelos por niño: {caramelos_por_niño}")
print(f"Caramelos sobrantes: {caramelos_sobrantes}")
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---

caramelos = 17
niños = 5

# División Entera: ¿Cuántas veces cabe 5 en 17?
caramelos_por_niño = caramelos // niños

# Módulo: ¿Cuánto sobra de la división 17 / 5?
caramelos_sobrantes = caramelos % niños

print(f"Caramelos por niño: {caramelos_por_niño}")  # Debería ser 3
print(f"Caramelos sobrantes: {caramelos_sobrantes}")  # Debería ser 2
```

**Explicación:**
* `//` (División Entera): Descarta la parte decimal de la división. `17 / 5` es `3.4`, por lo que `17 // 5` es `3`.
* `%` (Módulo): Devuelve el resto de la división. `5 * 3` es `15`. `17 - 15` es `2`.

**Nivel:** Avanzado

**Objetivo:** Entender que las variables creadas dentro de una función son *locales* a esa función.

**Consigna:**
1. Define una función simple `mi_funcion()`.
2. Dentro de la función, crea una variable `variable_local = "Soy local"`.
3. Llama a la función `mi_funcion()`.
4. Después de llamar a la función, intenta imprimir `variable_local` (fuera de la función).

```python
def mi_funcion():
    variable_local = "Soy local"
    print(f"Dentro de la función: {variable_local}")


mi_funcion()
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---


def mi_funcion():
    variable_local = "Soy local"
    print(f"Dentro de la función: {variable_local}")


# Llamamos a la función
mi_funcion()

# Este intento dará un error
try:
    print(variable_local)
except NameError as e:
    print(f"\nError al intentar acceder fuera: {e}")
```

**Explicación:**

* **Scope (Ámbito):** `variable_local` solo existe *dentro* del bloque de `mi_funcion()`.
* Una vez que la función termina de ejecutarse, esa variable es destruida. No es accesible desde el ámbito *global* (fuera de la función).

### 🔹 Ejercicio 18: Scope (Ámbito) Global (Lectura)
**Nivel:** Avanzado

**Objetivo:** Entender que las funciones pueden *leer* variables globales.

**Consigna:**
1. Crea una variable *global* `usuario_global = "Admin"`.
2. Define una función `saludar()`.
3. Dentro de `saludar()`, imprime la `usuario_global`.
4. Llama a la función `saludar()`.

```python
# Tu código aquí

usuario_global = "Admin"


def saludar():
    print(f"Hola!!! {usuario_global}")


saludar()
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---

# Variable en el ámbito Global
usuario_global = "Admin"


def saludar():
    # La función busca 'usuario_global' localmente, no la encuentra,
    # y la busca en el ámbito superior (global).
    print(f"Hola, {usuario_global}!")


saludar()
```

**Explicación:**
Esto funciona porque Python sigue la regla **LEGB** (Local -> Enclosing -> Global -> Built-in).
1. `saludar()` busca `usuario_global` en su ámbito **L**ocal. No la encuentra.
2. (No hay ámbito **E**nclosing aquí).
3. `saludar()` busca `usuario_global` en el ámbito **G**lobal. ¡La encuentra! Y la usa.

### 🔹 Ejercicio 20: Variables como Referencias (Mutabilidad)
**Nivel:** Avanzado

**Objetivo:** Comprender que las variables son "etiquetas" que apuntan a objetos en memoria, y la diferencia entre objetos mutables e inmutables.

**Consigna:**
1. Crea una lista `lista_a = [1, 2, 3]`.
2. Crea una variable `lista_b` asignándole `lista_a` (Ej: `lista_b = lista_a`).
3. Usando `lista_b`, añade el número 4 al final (Ej: `lista_b.append(4)`).
4. Imprime `lista_a`.
5. (Opcional) Imprime `id(lista_a)` e `id(lista_b)` para ver sus direcciones de memoria.

**Pregunta (reflexión):** ¿Por qué cambió `lista_a` si solo modificaste `lista_b`?

```python
lista_a = [1, 2, 3]
lista_b = lista_a

print(f"ID de lista_a: {id(lista_a)}")
print(f"ID de lista_b: {id(lista_b)}")
print(f"¿Son el mismo objeto? {lista_a is lista_b}")

print("\nModificando lista_b...")
lista_b.append(4)

print(f"Lista A (después): {lista_a}")
print(f"Lista B (después): {lista_b}")
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---

lista_a = [1, 2, 3]
lista_b = lista_a

print(f"ID de lista_a: {id(lista_a)}")
print(f"ID de lista_b: {id(lista_b)}")
print(f"¿Son el mismo objeto? {lista_a is lista_b}")

print("\nModificando lista_b...")
lista_b.append(4)

print(f"Lista A (después): {lista_a}")
print(f"Lista B (después): {lista_b}")
```

**Explicación:**
En Python, la asignación (`=`) no copia el objeto, **copia la referencia** (la "etiqueta" o "puntero").

1.  `lista_a = [1, 2, 3]`: Python crea un objeto (la lista) en la memoria y la etiqueta `lista_a` apunta a él.
2.  `lista_b = lista_a`: Python **no** crea una nueva lista. Simplemente crea una nueva etiqueta `lista_b` y la hace apuntar **al mismo objeto** al que apunta `lista_a`.
3.  `lista_b.append(4)`: Como `lista_b` apunta al objeto original, este método modifica ese objeto. Las listas son **mutables** (pueden cambiar).
4.  `print(lista_a)`: Muestra `[1, 2, 3, 4]`, porque `lista_a` sigue apuntando a ese mismo objeto, que ahora ha sido modificado.

**Diferencia con Inmutables:**
Si hicieras esto con un `int` (que es **inmutable**):
`a = 5`
`b = a`
`b = b + 1` (o `b += 1`)
Aquí, `a` seguiría valiendo `5` y `b` valdría `6`. ¿Por qué? Porque `b = b + 1` *no modifica* el objeto `5`. Crea un *nuevo* objeto `6` y hace que `b` apunte a él, mientras `a` sigue apuntando al objeto `5` original.
