### 2. Operaciones Aritméticas (int)

```python
# Completa tu código aquí
a = 15
b = 4

print(f"Suma (a + b): {a + b}")
print(f"Resta (a - b): {a - b}")
print(f"Multiplicación (a * b): {a * b}")
print(f"División (a / b): {a / b}")
print(f"División Entera (a // b): {a // b}")
print(f"Módulo (a % b): {a % b}")
print(f"Potencia (a ** b): {a ** b}")
```

### 4. Input de Usuario y Type Casting

    3.  ¿Tiene la persona una edad diferente a 25?
    4.  ¿Tiene la persona licencia?

* **Pista:** Usa `==` para "exactamente igual", `!=` para "diferente", y `>=` para "mayor o igual".

```python
# Completa tu código aquí
edad = 18
tiene_licencia = False

es_mayor = edad >= 18
print(f"Es mayor de edad?: {es_mayor}")

tiene_18 = edad == 18
print(f"Tiene 18 años?: {tiene_18}")

es_diferente_25 = edad != 25
print(f"Edad es diferente a la edad de 25: {es_diferente_25}")

print(f"Tiene licencia?: {tiene_licencia}")
```

### 7. Casting de float a int

* **Nivel:** Básico
* **Objetivo:** Entender cómo se pierde información al convertir de `float` a `int`.
* **Consigna:** Tienes `precio_decimal = 19.99`. Convierte este precio a un entero usando `int()` y guárdalo en `precio_entero`. Imprime ambas variables y sus tipos.
* **Pregunta:** ¿Qué le pasó a la parte decimal?

```python
# Completa tu código aquí
precio_decimal = 19.99
precio_entero = int(precio_decimal)

print(f"precio decimal: {precio_decimal} (Tipo: {type(precio_decimal)})")
print(f"precio entero: {precio_entero} (Tipo: {type(precio_entero)})")
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---
precio_decimal = 19.99
precio_entero = int(precio_decimal)

print(f"Precio decimal: {precio_decimal} (Tipo: {type(precio_decimal)})")
print(f"Precio entero: {precio_entero} (Tipo: {type(precio_entero)})")
```

### 9. Métodos de String: `.upper()`, `.lower()`, `.strip()`

* **Nivel:** Intermedio
* **Objetivo:** Limpiar y normalizar un string.
* **Consigna:** Tienes un string con formato inconsistente: `entrada_usuario = "  HoLa MuNdO  "`. 
    Realiza los siguientes pasos:
    1.  Crea `limpio` eliminando los espacios en blanco del principio y el final.
    2.  Crea `mayusculas` convirtiendo `limpio` todo a mayúsculas.
    3.  Crea `minusculas` convirtiendo `limpio` todo a minúsculas.
    Imprime las tres nuevas variables.

* **Pista:** Los métodos se llaman con un punto (ej: `variable.metodo()`).

```python
# Completa tu código aquí
entrada_usuario = "  HoLa MuNdO  "

limpio = entrada_usuario.strip()
mayusculas = limpio.upper()
minusculas = limpio.lower()

print(f"Original: '{entrada_usuario}'")
print(f"Limpio: '{limpio}'")
print(f"Mayúsculas: '{mayusculas}'")
print(f"Minúsculas: '{minusculas}'")
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---
entrada_usuario = "  HoLa MuNdO  "

# 1. Quitar espacios
limpio = entrada_usuario.strip()

# 2. Convertir a mayúsculas
mayusculas = limpio.upper()

# 3. Convertir a minúsculas
minusculas = limpio.lower()

print(f"Original: '{entrada_usuario}'")
print(f"Limpio: '{limpio}'")
print(f"Mayúsculas: '{mayusculas}'")
print(f"Minúsculas: '{minusculas}'")
```

#### Explicación
* **Qué hace el código:** `.strip()` elimina espacios, tabulaciones y saltos de línea del inicio y el final del string. `.upper()` devuelve una nueva copia del string en mayúsculas. `.lower()` devuelve una nueva copia en minúsculas.
* **Errores comunes:** Pensar que los métodos modifican el string original. Los strings en Python son **inmutables**. `entrada_usuario.strip()` *devuelve* un nuevo string, no modifica `entrada_usuario`. Por eso debemos asignar el resultado a una nueva variable.

### 10. Indexing de Strings (Acceder a caracteres)

* **Nivel:** Intermedio
* **Objetivo:** Obtener caracteres específicos de un string usando su índice.

### 11. Slicing de Strings (Rebanadas)

* **Nivel:** Intermedio
* **Objetivo:** Obtener sub-secciones (rebanadas) de un string.
* **Consigna:** Dado el string `url = "https://www.google.com"`:
    1.  Obtén el protocolo (los primeros 5 caracteres).
    2.  Obtén el dominio (desde el carácter 8 hasta el 17).
    3.  Obtén la extensión (los últimos 4 caracteres).

* **Pista:** La sintaxis es `[inicio:fin]`. `inicio` es inclusivo, `fin` es exclusivo. Si omites `inicio`, empieza desde 0. Si omites `fin`, va hasta el final.

```python
# Completa tu código aquí
url = "https://www.google.com"

protocolo = url[0:5]
print(f"Protocolo: {protocolo}")
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---
url = "https://www.google.com"

# 1. Protocolo (índices 0, 1, 2, 3, 4)
protocolo = url[0:5]  # o [:5]
print(f"Protocolo: {protocolo}")

# 2. Dominio (índices 8 al 17)
dominio = url[8:18]  # 'google.com'
print(f"Dominio: {dominio}")

# 3. Extensión (últimos 4)
extension = url[-4:]
print(f"Extensión: {extension}")
```

#### Explicación
* **Qué hace el código:** El slicing `[inicio:fin]` extrae caracteres.
    * `[:5]` -> Desde el inicio (índice 0) hasta (sin incluir) el índice 5.
    * `[8:18]` -> Desde el índice 8 hasta (sin incluir) el índice 18.
    * `[-4:]` -> Desde el 4to carácter contando desde el final (índice -4) hasta el final del string.

### 12. Operadores Booleanos (`and`, `or`, `not`)

* **Nivel:** Intermedio
* **Objetivo:** Combinar múltiples condiciones booleanas.
* **Consigna:** Simula las condiciones para entrar a una atracción. 
    `edad = 14`
    `altura_cm = 160`
    `tiene_pase_vip = False`
    
    Evalúa e imprime las siguientes condiciones:
    1.  ¿Es mayor de 12 años **y** mide más de 150 cm?
    2.  ¿Es mayor de 18 años **o** mide más de 170 cm?
    3.  ¿Puede entrar gratis? (Solo si tiene pase VIP).
    4.  ¿No tiene pase VIP? (Usar `not`).

```python
# Completa tu código aquí
edad = 14
altura_cm = 160
tiene_pase_vip = False
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---
edad = 14
altura_cm = 160
tiene_pase_vip = False

# 1. ¿Es mayor de 12 años Y mide más de 150 cm?
condicion_1 = (edad > 12) and (altura_cm > 150)
print(f"Condición 1 (and): {condicion_1}")

# 2. ¿Es mayor de 18 años O mide más de 170 cm?
condicion_2 = (edad > 18) or (altura_cm > 170)
print(f"Condición 2 (or): {condicion_2}")

# 3. ¿Puede entrar gratis? (Solo si tiene pase VIP)
condicion_3 = tiene_pase_vip
print(f"Condición 3 (pase): {condicion_3}")

# 4. ¿No tiene pase VIP? (Usar not)
condicion_4 = not tiene_pase_vip
print(f"Condición 4 (not): {condicion_4}")
```

#### Explicación
* **`and`:** Devuelve `True` solo si AMBAS condiciones son `True`. (En el ej. 1: `True and True` -> `True`).
* **`or`:** Devuelve `True` si AL MENOS UNA condición es `True`. (En el ej. 2: `False or False` -> `False`).
* **`not`:** Invierte el valor booleano. `not True` es `False`, y `not False` es `True`.

    Reemplaza `"[NOMBRE]"` por `"Ana"` y `"[CIUDAD]"` por `"Barcelona"` para crear un saludo personalizado.

### 17. Formateo Avanzado de f-strings (float)

* **Nivel:** Avanzado
* **Objetivo:** Usar f-strings para controlar el formato de números flotantes (precisión decimal).
* **Consigna:** Tienes una variable `pi = 3.14159265`. Imprime el valor de PI formateado de las siguientes maneras:
    1.  Con solo 2 decimales.
    2.  Con solo 4 decimales.
    3.  En un espacio total de 10 caracteres, alineado a la derecha, con 2 decimales.

* **Pista:** La sintaxis es `{variable:.Nf}` para N decimales. Para alineación: `{variable:[ancho].[precision]f}`.

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---
pi = 3.14159265

# 1. Con 2 decimales
print(f"PI (2 decimales): {pi:.2f}")

# 2. Con 4 decimales
print(f"PI (4 decimales): {pi:.4f}")

# 3. Ancho 10, alineado derecha, 2 decimales
print(f"PI (formateado): {pi:>10.2f}")
```

### 18. Slicing Avanzado (step)

* **Nivel:** Avanzado
* **Objetivo:** Usar el parámetro `step` (paso) en el slicing de strings.
* **Consigna:** Dada la frase `texto = "0a1b2c3d4e5f6g7h8i9j"`
    1.  Extrae solo los números (empezando en el índice 0, saltando de 2 en 2).
    2.  Extrae solo las letras (empezando en el índice 1, saltando de 2 en 2).
    3.  Invierte el string completo.

* **Pista:** La sintaxis es `[inicio:fin:paso]`. Un paso de `-1` invierte el string.

```python
# Completa tu código aquí
texto = "0a1b2c3d4e5f6g7h8i9j"
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---
texto = "0a1b2c3d4e5f6g7h8i9j"

# 1. Solo números (inicio 0, paso 2)
numeros = texto[::2]
print(f"Números: {numeros}")

# 2. Solo letras (inicio 1, paso 2)
letras = texto[1::2]
print(f"Letras: {letras}")

# 3. Invertir el string (paso -1)
invertido = texto[::-1]
print(f"Invertido: {invertido}")
```

#### Explicación
* **Qué hace el código:**
    * `[::2]`: Omite `inicio` y `fin` (toma todo el string), pero usa un `paso` de 2. Toma los índices 0, 2, 4, ...
    * `[1::2]`: Empieza en el índice 1, toma todo el resto, con un `paso` de 2. Toma los índices 1, 3, 5, ...
    * `[::-1]`: Un `paso` negativo de -1 recorre el string al revés. Es el truco más común en Python para invertir un string.

* **Nivel:** Avanzado
* **Objetivo:** Entender qué valores (no booleanos) evalúa Python como `True` o `False`.
* **Consigna:** Usa la función `bool()` para determinar el valor de verdad (Truthiness) de los siguientes elementos:
    1.  `valor_int_cero = 0`
    2.  `valor_int_positivo = 42`
    3.  `valor_float_cero = 0.0`
    4.  `valor_string_vacio = ""`
    5.  `valor_string_lleno = "Hola"`
    6.  `valor_none = None` (El tipo `NoneType`, que representa la ausencia de valor)

* **Pista:** Generalmente, los "ceros" o "vacíos" de cada tipo se evalúan como `False`.

```python
# Completa tu código aquí
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---
valor_int_cero = 0
valor_int_positivo = 42
valor_float_cero = 0.0
valor_string_vacio = ""
valor_string_lleno = "Hola"
valor_none = None

print(f"bool(0): {bool(valor_int_cero)}")
print(f"bool(42): {bool(valor_int_positivo)}")
print(f"bool(0.0): {bool(valor_float_cero)}")
print(f"bool(\\"\\"): {bool(valor_string_vacio)}")
print(f"bool(\"Hola\"): {bool(valor_string_lleno)}")
print(f"bool(None): {bool(valor_none)}")
```

#### Explicación

* **Punto clave (Valores "Falsy"):** En Python, los siguientes valores se consideran `False`:
    * `False`
    * `None`
    * `0` (int)
    * `0.0` (float)
    * `""` (string vacío)
    * `[]` (lista vacía), `{}` (diccionario vacío), `()` (tupla vacía)
* Cualquier otro valor (como `42` o `"Hola"`) se considera **"Truthy"** (evalúa como `True`).

### 20. Encadenamiento de Métodos (Method Chaining)

* **Nivel:** Avanzado
* **Objetivo:** Combinar múltiples métodos de string en una sola línea.
* **Consigna:** Tienes el siguiente string: `datos_sucios = "   ID_Usuario: 12345, Nombre: Juan   "`. 
    En una sola línea de código (asignada a `datos_limpios`), realiza las siguientes 3 acciones:
    1.  Elimina los espacios en blanco de los extremos.
    2.  Pon todo el texto en minúsculas.
    3.  Reemplaza el `"id_usuario: "` (nota el espacio) por `"id: "`.

    Imprime `datos_limpios`.

* **Pista:** `variable.metodo1().metodo2().metodo3()`

```python
# Completa tu código aquí
datos_sucios = "   ID_Usuario: 12345, Nombre: Juan   "
```

```python
# --- SOLUCIÓN (mostrar solo si el usuario lo pide) ---
datos_sucios = "   ID_Usuario: 12345, Nombre: Juan   "

# 1. .strip() -> 2. .lower() -> 3. .replace()
datos_limpios = datos_sucios.strip().lower().replace("id_usuario: ", "id: ")

print(f"Sucios: '{datos_sucios}'")
print(f"Limpios: '{datos_limpios}'")
```

#### Explicación
* **Qué hace el código:** Esto es posible porque cada método (`.strip()`, `.lower()`, `.replace()`) devuelve un *nuevo string*, sobre el cual se puede llamar inmediatamente al siguiente método.
* **Orden de ejecución:**
    1.  `datos_sucios.strip()` -> `"ID_Usuario: 12345, Nombre: Juan"`
    2.  *resultado_1*.`lower()` -> `"id_usuario: 12345, nombre: juan"`
    3.  *resultado_2*.`replace(...)` -> `"id: 12345, nombre: juan"`
* **Errores comunes:** El orden importa. Si hicieras `.lower()` antes que `.strip()`, funcionaría igual, pero si hicieras `.replace()` primero, fallaría, porque `replace` busca `"id_usuario: "` (minúsculas) y el string original aún estaría en mayúsculas (`"ID_Usuario: "`).
